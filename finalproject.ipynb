```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
import os

data=pd.read_excel(r"/content/Sample - Superstore_uncleaned data (1).xlsx")
data
```

```python
data.describe()
```

```python
data.isnull().sum()
```

```python
data.info()
```

```python
data['Ship Mode'].fillna(value="unknow", inplace=True)
data['Ship Mode'].unique()
```

```python
data["Country"].fillna(value='United States', inplace=True)
data["Country"].unique()
```

```python
data['State'].fillna(value='unknow',inplace=True)
data['State'].unique()
```

```python
data['Region']=data['Region'].replace({'E':'East','w':'West'})
data['Region'].fillna(value='West', inplace=True)
data['Region'].unique()
```

```python
data['Segment']=data['Segment'].replace({'HO':'Home Office','Cor':'Corporate'})
data['Segment'].unique()
```

#KPI Requirement


```python
Total_sales=data['Sales'].sum()
Total_profit=data['Profit'].sum()
Total_quantity=data['Quantity'].sum()
Total_amount=data['COST PRICE'].sum()
net_profit=(data['Profit']/data['Sales']*100).mean()

print(f"Total sales={Total_sales}")
print(f"Total profit={Total_profit}")
print(f"Total quantity={Total_quantity}")
print(f"Total amount={Total_amount}")
print(f"Net profit precent={net_profit}%")
```

# Monthly sales trend to spot seasonality

```python
data['Month']=data['Order Date'].dt.month_name()
Trends_over_sales=data.groupby('Month')['Sales'].sum()
month_order = ['January','February','March','April','May','June',
               'July','August','September','October','November','December']
Trends_over_sales = Trends_over_sales.reindex(month_order)
plt.figure(figsize=(12,6))
plt.plot(Trends_over_sales.index, Trends_over_sales.values)
plt.xlabel("Order Date")
plt.ylabel("Total Sales")
plt.title("Trends over Sales")
plt.show()

```

#Sales by category

```python
sales_by_category=data.groupby('Category')['Sales'].sum()
fig, ax = plt.subplots(figsize=(6,6))
wedges, texts, autotexts = ax.pie(sales_by_category, labels=sales_by_category.index, autopct='%1.1f%%', startangle=90, pctdistance=0.85)
centre=plt.Circle((0,0),0.75,color='white')
fig.gca().add_artist(centre)
plt.title('sales by category')
plt.tight_layout()
plt.show()
```

#Sales by Sub-category

```python
sales_by_subcategory=data.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False)
plt.figure(figsize=(10,6))
sns.barplot(x=sales_by_subcategory.index, y=sales_by_subcategory.values, palette='viridis')
plt.xticks(rotation=90, ha='center')
plt.title('sales by sub-category')
plt.ylabel('Sales')
plt.tight_layout()
```

#Sales in top 10 state

```python
top_state=data.groupby('State')['Sales'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,5))
sns.barplot(x=top_state.index, y=top_state.values, palette='plasma')
plt.xticks(rotation=60, ha='center')
plt.xlabel('Total sales')
plt.ylabel('State')
plt.title("Top 10 state sales")
plt.tight_layout()
plt.show()
```

#Sales share by region

```python
sale_region=data.groupby('Region')['Sales'].sum().sort_values(ascending=False)
fig,ax=plt.subplots(figsize=(6,6))
wedges,texts,autotexts=ax.pie(sale_region,labels=sale_region.index, autopct="%1.1f%%", startangle=90,pctdistance=0.81)
plt.title('Sales share by region')
```

```python
grouped=data.groupby(['Segment','Category'])['Sales'].sum().unstack('Category')
grouped.plot(kind='bar', figsize=(10,6))
plt.xlabel('segment')
plt.ylabel('sales')
plt.title('Sales by Segment and category')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#All Matrix by outlet type

```python
grouped_metrics = data.groupby('Category').agg({
    'Sales': ['sum', 'mean', 'count'],
    'Profit': 'mean'  })

grouped_metrics.columns = ['Total Sales', 'Average Sales', 'Number of Items', 'Average profit']
grouped_metrics = grouped_metrics.reset_index()

print("\n Matrix Card: Key Metrics by Outlet Type")
print(grouped_metrics)
```
